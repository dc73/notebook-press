$$y = \sin(x) e^{-x/8}$$

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 12, 300)
y = np.sin(x) * np.exp(-x / 8)

fig, ax = plt.subplots(figsize=(7, 3.6), layout="constrained")
ax.plot(x, y, color="#52745b", linewidth=2.5)
ax.axhline(0, color="#cbd6c7", linewidth=1)
ax.set(title="Damped oscillation", xlabel="Time", ylabel="Amplitude")
ax.spines[["top", "right"]].set_visible(False)
plt.show()
```
